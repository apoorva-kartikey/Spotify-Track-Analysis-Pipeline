```python
# Spotify Track Popularity Analysis
## Notebook 02: Exploratory Data Analysis (EDA)

#Objective:** Explore the structural properties of our 10,000-row sampled dataset. We will analyze target distribution, track audio attribute relationships, and isolate top-performing artists and genres to extract actionable baseline insights before modeling.
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Set clean visualization themes for portfolio consistency
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['axes.titlesize'] = 14
plt.rcParams['axes.labelsize'] = 12

# ==========================================================
# 2. Load the Sampled Dataset
# ==========================================================
data_path = r"C:\Users\SUJATA\Desktop\Data\Data Science\spotify_sample_10000.csv"
df = pd.read_csv(data_path)

print(f"Dataset successfully loaded for EDA. Shape: {df.shape}")
```

```python
# Create a blank canvas
plt.figure(figsize=(10, 5))

# Draw a histogram with a smooth trend line (KDE = Kernel Density Estimate)
sns.histplot(df['popularity'], bins=30, kde=True, color='#1DB954') 

plt.title('Distribution of Spotify Song Popularity Scores')
plt.xlabel('Popularity Score (0-100)')
plt.ylabel('Track Count')

# Saving the plot to a specified directory with high resolution
plt.savefig(r"C:\Users\SUJATA\Desktop\Data\Data Science\popularity_distribution.png", dpi=300)
plt.show()
```

```python
audio_features = [
    'popularity', 'danceability', 'energy', 'loudness', 'liveness'
]

# Filter the dataset to keep only these specific audio columns
df_numerical = df[audio_features]

# Set the chart canvas size explicitly
plt.figure(figsize=(8, 6))

# Draw the heatmap plot with labels and a warm-cool color scale
sns.heatmap(
    df_numerical.corr(), 
    annot=True,      
    cmap='coolwarm', 
    fmt=".2f",       
    linewidths=0.5   
)

plt.title('Correlation Matrix of Selected Audio Metrics')
plt.tight_layout()

# Save the image to the local project folder
plt.savefig(r"C:\Users\SUJATA\Desktop\Data\Data Science\correlation_heatmap.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
top_artists = df['artists'].value_counts().head(10)

sns.barplot(x=top_artists.values, y=top_artists.index, palette='viridis')
plt.title('Top 10 Most Frequent Artists in Dataset')
plt.xlabel('Total Tracks Contributed')
plt.ylabel('Artist Name')
plt.tight_layout()

plt.savefig(r"C:\Users\SUJATA\Desktop\Data\Data Science\top_artists.png", dpi=300)
plt.show()
```
